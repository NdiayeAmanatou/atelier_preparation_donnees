##### 12) h) normaliser les catégories textuelles en supprimant les espaces puis en uniformisant la casse 

```python
colonnes_a_normaliser = ["batiment","type_batiment","zone","mode_climatisation","etat_systeme","jour_semaine","alerte"]

for col in colonnes_a_normaliser:
    df[col] = df[col].str.strip().str.lower()
```

```python
for col in colonnes_a_normaliser:
    print(f"--- {col} ---")
    print(sorted(df[col].dropna().unique()))
    print(df[col].nunique(), "modalités")
    print()
```

correction des fautes de frappe 

```python
corrections_type_bat = {
    "bureu": "bureau",
    "ecole": "école",
    "entrepot": "entrepôt"
}
corrections_mode_clim = {
    "normale": "normal"
}

df["type_batiment"] = df["type_batiment"].replace(corrections_type_bat)
df["mode_climatisation"] = df["mode_climatisation"].replace(corrections_mode_clim)
```

```python
for col in colonnes_a_normaliser:
    print(f"--- {col} ---")
    print(sorted(df[col].dropna().unique()))
    print(df[col].nunique(), "modalités")
    print()
```

##### 13) c) Quelle stratégie utiliser pour les valeurs manquantes ?

La strategie a utiliser pour les valeurs manquantes est soit on les supprime s'ils ne sont pas trop influentes, oubien on les remplace par soit la moyenne, la mediane ou le mode c'est a dire la valeur la plus frequente pour les variables categorielles.

##### 13) d) Peut-on supprimer toutes les lignes contenant des valeurs manquantes ? 

Non on ne peut supprimer toutes les lignes contenant les valeurs manquantes . ca depend des conditions par exemple si on a :

--très peu de lignes concernées (moins de 5 %)

--manquants répartis au hasard

--grand jeu de données, où perdre quelques lignes ne change rien

##### 13) e) Dans quels cas utiliser la moyenne ? 

On utilise la moyenne pour remplacer des valeurs manquantes lorsque la variable est numérique et continue, que sa distribution est à peu près symétrique et qu'elle ne contient pas de valeurs aberrantes importantes, car la moyenne y est très sensible. Le pourcentage de manquants doit aussi rester faible, sinon l'imputation réduit artificiellement la variabilité des données.

##### 13) f) Quand préférer la médiane ?

on préfère la médiane quand la variable est numérique et qu'elle contient des valeurs aberrantes ou une distribution asymétrique, car elle est robuste aux extrêmes, contrairement à la moyenne.

##### 13) g) Comment traiter une variable catégorielle ?

on traite les manquants d'une variable catégorielle par le mode, par une modalité « inconnu », ou par déduction depuis d'autres colonnes, mais jamais par la moyenne ou la médiane.
