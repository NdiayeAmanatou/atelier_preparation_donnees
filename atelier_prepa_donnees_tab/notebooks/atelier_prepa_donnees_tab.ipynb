##### 11) Y a-t-il des variables potentiellement problématiques 

Oui il y'a plusieurs variables potentiellement problematiques .
--on observe qu'il y'a des valeurs manquantes dans toutes les variables.

id_mesure a un count de 507, alors que les autres colonnes ont entre 495 et 502 valeurs non nulles. Donc il manque environ 12 valeurs en température, 11 en humidité, 7 en CO2, 6 en occupation et 5 en consommation.

--On a probablement la presences de doublons sur id_mesure.Les identifiants vont de 1001 à 1500, soit 500 valeurs possibles, mais il y a 507 lignes. Il y a donc probablement des lignes en double.

--On constate egalement des valeurs aberrantes:

temperature : de -30 à 96 °C, alors que 50 % des valeurs sont entre 21,6 et 26. Ces extrêmes ressemblent à des erreurs de capteur.

humidite : de -12 à 160, alors qu'une humidité en % doit rester entre 0 et 100.

co2 : le max de 6000 est très loin du 75e percentile (952), et l'écart-type (582) est énorme par rapport à la médiane. Le min de 89 est aussi suspect, car l'air extérieur est déjà autour de 400 ppm.

occupation : le min de -20 est impossible. car on ne peut pas avoir un nombre negatif de personnes.

consommation_kwh : le min de -100 est impossible (une consommation ne peut pas être négative).

##### 12) f) Si une valeur est manifestement erronée et qu’on ne peut pas retrouver sa vraie valeur, la transformer en valeur manquante 

```python
nan_avant = df[["humidite", "consommation_kwh", "occupation"]].isna().sum()
print(nan_avant)
```

```python
df.loc[(df["humidite"] < 0) | (df["humidite"] > 100), "humidite"] = np.nan
```

verification

```python
print(df["humidite"].isna().sum())
```

```python
df.loc[(df["consommation_kwh"] < 0), "consommation_kwh"] = np.nan
```

verification

```python
print(df["consommation_kwh"].isna().sum())
```

```python
df.loc[(df["occupation"] < 0), "occupation"] = np.nan
```

verification

```python
print(df["occupation"].isna().sum())
```
