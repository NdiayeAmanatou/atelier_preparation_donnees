##### 13) e) Dans quels cas utiliser la moyenne ? 

On utilise la moyenne pour remplacer des valeurs manquantes lorsque la variable est numérique et continue, que sa distribution est à peu près symétrique et qu'elle ne contient pas de valeurs aberrantes importantes, car la moyenne y est très sensible. Le pourcentage de manquants doit aussi rester faible, sinon l'imputation réduit artificiellement la variabilité des données.

##### 13) f) Quand préférer la médiane ?

on préfère la médiane quand la variable est numérique et qu'elle contient des valeurs aberrantes ou une distribution asymétrique, car elle est robuste aux extrêmes, contrairement à la moyenne.

##### 13) g) Comment traiter une variable catégorielle ?

on traite les manquants d'une variable catégorielle par le mode, par une modalité « inconnu », ou par déduction depuis d'autres colonnes, mais jamais par la moyenne ou la médiane.

### Partie 4 – Encodage des variables catégorielles 

##### Appliquer, en justifiant, l’encodage le plus adapté à chaque variable catégorielle 

--pour batiment, type_batiment, zone, jour_semaine on applique le One-Hot . car ces variables la sont nominale ils ne suivent pas une ordre. 

--pour etat_systeme, mode_climatisation : on applique le Ordinal encoding car les variables sont ordonnees.

--pour alerte (la cible y) : on applique le binary encoding car ici on a que deux classes .


```python
from sklearn.preprocessing import OneHotEncoder
import pandas as pd

cols_onehot = ["batiment","type_batiment","zone","jour_semaine"]

# Création de l'encodeur
encodeur = OneHotEncoder(handle_unknown="ignore", sparse_output=False)

# Apprentissage sur le train, appliquer au test
train_encode = encodeur.fit_transform(X_train[cols_onehot])
test_encode = encodeur.transform(X_test[cols_onehot])

# Remet des noms de colonnes et garder les index d'origine
noms = encodeur.get_feature_names_out(cols_onehot)
train_encode = pd.DataFrame(train_encode, columns=noms, index=X_train.index)
test_encode = pd.DataFrame(test_encode, columns=noms, index=X_test.index)

# Remplace les anciennes colonnes par les nouvelles
X_train = pd.concat([X_train.drop(columns=cols_onehot), train_encode], axis=1)
X_test = pd.concat([X_test.drop(columns=cols_onehot), test_encode], axis=1)
```

```python
print(X_train.shape, X_test.shape)
print(X_train.columns.tolist())
print(X_train.head())
```

**Note :** pour l'encodage de etat_systeme, mode_climatisation avec Ordinal-encoding

```python
from sklearn.preprocessing import OrdinalEncoder

cols_ord = ["etat_systeme", "mode_climatisation"]

# Ordre des modalités, du plus petit au plus grand (même ordre que cols_ord)
ordre_etat = ["normal", "alerte", "panne"]
ordre_mode = ["eco", "normal", "boost"]

encodeur_ord = OrdinalEncoder(
    categories=[ordre_etat, ordre_mode],
    handle_unknown="use_encoded_value",
    unknown_value=-1
)

# Apprendre sur le train, appliquer au test
X_train[cols_ord] = encodeur_ord.fit_transform(X_train[cols_ord])
X_test[cols_ord] = encodeur_ord.transform(X_test[cols_ord])
```

```python
# Variables numériques continues -> médiane
cols_mediane = [c for c in ["temperature", "humidite", "co2", "occupation", "consommation_kwh"]
                if c in X_train.columns]

# Toutes les autres colonnes (ordinales, encodées...) -> valeur la plus fréquente
cols_autres = [c for c in X_train.columns if c not in cols_mediane]

# Valeurs de remplacement calculées sur le TRAIN uniquement
medianes = X_train[cols_mediane].median()
modes = X_train[cols_autres].mode().iloc[0]
valeurs_remplacement = pd.concat([medianes, modes])

# Appliquer au train ET au test
X_train = X_train.fillna(valeurs_remplacement)
X_test = X_test.fillna(valeurs_remplacement)
```

### Partie 5 – Mise à l'échelle 

##### 1) A l’aide de quelques variables, justifier la pertinence de la mise à l’échelle 

***Justification de la mise a l'echelle***

Les variables ne sont pas exprimées dans les mêmes unités ni au même ordre de grandeur. Le CO₂ (en ppm) a une moyenne d'environ 844 et un écart-type d'environ 582, avec des valeurs de 89 à 6000. La température (en °C) a une moyenne d'environ 24 et un écart-type d'environ 7. L'humidité (en %) a une moyenne d'environ 58 (écart-type 16), et la consommation (en kWh) d'environ 169 (écart-type 53).

Sans mise à l'échelle, les algorithmes basés sur les distances (KNN, K-means, SVM) traiteraient un écart de 200 ppm de CO₂ comme bien plus important qu'un écart de 5 °C, uniquement à cause de l'unité et non de l'information apportée. Le CO₂ dominerait le calcul. Les modèles à gradient (régression, réseaux de neurones) convergent aussi plus lentement, et leurs coefficients sont difficiles à comparer.

La mise à l'échelle ramène les variables à des ordres de grandeur comparables, pour que chacune contribue de façon équitable. Les arbres de décision et les forêts aléatoires n'en ont pas besoin, car ils ne comparent que des seuils sur une seule variable à la fois.

##### 2) Appliquer et justifier une technique de mise à l'échelle 

```python
from sklearn.preprocessing import StandardScaler

cols_num = ["temperature", "humidite", "co2", "occupation", "consommation_kwh"]

scaler = StandardScaler()

# Apprentissage sur le train, application au train et au test
X_train[cols_num] = scaler.fit_transform(X_train[cols_num])
X_test[cols_num] = scaler.transform(X_test[cols_num])
```

***Verifiacation***

```python
print(X_train[cols_num].agg(["mean", "std"]).round(2))
print(X_test[cols_num].agg(["mean", "std"]).round(2))
print(X_train.shape, X_test.shape)
```

***Justification***: 

J'ai utilisé la standardisation (StandardScaler), qui transforme chaque variable avec la formule (x − moyenne) / écart-type. Chaque variable a alors une moyenne de 0 et un écart-type de 1, donc les variables deviennent comparables malgré leurs unités et leurs ordres de grandeur différents (CO₂ en ppm, température en °C, consommation en kWh).

Cette technique convient bien aux algorithmes sensibles à l'échelle (KNN, SVM, régression logistique, réseaux de neurones), et elle conserve la forme de la distribution, contrairement à la normalisation Min-Max qui ramène tout dans [0, 1]. Elle est adaptée quand les distributions sont à peu près symétriques, ce qui est le cas de [tes variables, d'après tes histogrammes et ton skew de la question 15].

Le scaler est appris (fit) sur le jeu d'entraînement uniquement, puis appliqué (transform) au jeu de test, pour éviter toute fuite d'information. Seules les variables numériques continues sont standardisées : les colonnes One-Hot, les colonnes ordinales et la cible y sont laissées telles quelles.
